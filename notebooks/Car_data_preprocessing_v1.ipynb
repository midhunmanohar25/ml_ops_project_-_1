## Analysing And Cleaning the data column wise

### Taking name column

```python
df['name'].unique()
```

```python
df.shape
```

```python
df = df[~(df['name'].str.startswith('Any')| df['name'].str.startswith('I') | df['name'].str.startswith('i') | df['name'].str.startswith('Commercial') | df['name'].str.endswith('Ta') | df['name'].str.startswith('all'))]
```

```python
df.shape
```

```python
df['name']= df['name'].str.split().str.slice(start=0, stop=3).str.join(' ')
df['name']
```

```python
df.name.unique()
```

#### It Still contain some rubbish names so try to find those names

```python
df[~df['name'].str.startswith(('Jaguar', 'Volvo', 'Land', 'Jeep', 'Force', 'Mercedes', 'Mini','Mitsubishi', 'Datsun','Fiat','Hyundai', 'Mahindra', 'Maruti', 'Ford', 'Skoda', 'Audi', 'Renault', 'Honda', 'Chevrolet', 'Volkswagen', 'Toyota', 'BMW', 'Hindustan', 'Nissan', 'Tata'))]
```

```python
df = df[df['name'].str.startswith(('Jaguar', 'Volvo', 'Land', 'Jeep', 'Force', 'Mercedes', 'Mini','Mitsubishi', 'Datsun','Fiat','Hyundai', 'Mahindra', 'Maruti', 'Ford', 'Skoda', 'Audi', 'Renault', 'Honda', 'Chevrolet', 'Volkswagen', 'Toyota', 'BMW', 'Hindustan', 'Nissan', 'Tata'))]
```

```python
df['name'].unique()
```

### Taking company column

```python
df['company'].value_counts()
```

```python
df = df[~df['company'].isin(['sell', 'Sale','Used','URJENT','very','Well','all','scratch','urgent','sell','Swift'])]
```

```python
df= df[~df['company'].str.isnumeric()]
```

```python
df.shape
```

```python
df['company'] = df['company'].str.lower()
```

```python
df['company'] = df['company'].str.capitalize()
```

```python
df.head()
```

### Taking year column

```python
df.year.dtype
# Here you can see the year column is in object datatype so we have to change it
```

```python
df.year.unique()
```

```python
df = df[df['year'].str.isnumeric()]
```

```python
df['year'] = df['year'].astype('int')
```

```python
df.shape
```

```python
df.head()
```

### Taking Price column

```python
df['Price'].unique()
```

```python
df['Price'].isnull().sum()
```

```python
df['Price'] = df['Price'].replace('Ask For Price', np.nan)

df['Price'] = (
    df['Price']
    .str.replace(',', '', regex=False)
    .astype('Int64')
)
```

```python
# import seaborn as sns
# sns.distplot(df['Price'])
```

```python
df.info()
```

### Taking kms_driven column

```python
df['kms_driven']
```

```python
df = df[~df['kms_driven'].str.isalpha()]
```

```python
df['kms_driven'] = df['kms_driven'].str.split(' ').str.get(0).str.replace(',','').astype('int')
```

```python
df.info()
```

### Taking fule_type column

```python
df['fuel_type'].unique()
```

```python
df['fuel_type'].isnull().sum()
```

```python
df['fuel_type'] = df['fuel_type'].fillna(value=df['fuel_type'].mode()[0])
```
